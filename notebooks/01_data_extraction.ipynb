# 01 - Data Extraction

## AdventureWorksDW2025 Analytics Project

This notebook extracts sales data from SQL Server
(AdventureWorksDW2025) and prepares the raw dataset
for the analytics pipeline.

### Objectives

- Connect to SQL Server
- Read the cleaned analytics table
- Validate the extracted structure
- Inspect the dataset
- Export the raw CSV dataset

```python
import sys

print(sys.executable)
```

```python
import os
import pandas as pd

from sqlalchemy import create_engine
from dotenv import load_dotenv
from urllib.parse import quote_plus

print("Libraries imported successfully.")
```

```python
load_dotenv()

server = os.getenv("SERVER")
database = os.getenv("DATABASE")
driver = os.getenv("DRIVER")

print("Server:", server)
print("Database:", database)
print("Driver:", driver)
```

```python
connection_string = (
    f"mssql+pyodbc://@{server}/{database}"
    f"?driver={quote_plus(driver)}"
    "&trusted_connection=yes"
)

engine = create_engine(connection_string)

print("SQL Server connection created successfully.")
```

```python
query = """
SELECT *
FROM analytics.Sales_Data_Cleaned
"""

df = pd.read_sql(
    query,
    engine
)

print("Data extracted successfully.")
```

```python
print("Shape:", df.shape)
```

```python
print("Columns:")
print(df.columns.tolist())
```

```python
print("Missing values:")
print(df.isna().sum())
```

```python
os.makedirs("../data/raw", exist_ok=True)

output_file = "../data/raw/sales_data.csv"

df.to_csv(
    output_file,
    index=False
)

print("Raw dataset exported:")
print(output_file)
```
